```python
#import library as name
import pandas as pd
```

```python
df = pd.read_csv("/content/Student_Performance_100_Notclean.csv")
df.head(7)
```

```python
df.info()
```

```python
subjects_cols = ["Python","Mathematics","DBMS","AI","Datastructures"]
pass_mark = 40
```

```python
df.dropna(inplace = True)
df.isnull().sum()
```

```python
print("Numbers of duplicate rows:",df.duplicated().sum())
```

```python
df.drop_duplicates(inplace = True)
print("Numbers of duplicate rows after dropping:",df.duplicated().sum())
```

create columns for total and average marks


```python
cols = ["Python", "Mathematics", "DBMS", "AI", "DataStructures"]
df["Total"] = df[cols].sum(axis=1)
df["Average"] = df[cols].mean(axis=1)
df.head()
```

```python
overall_average = df['Average'].mean()
print("The average of the average table is:", round(overall_average))
```

Identify the topper


```python
topper_index = df["Total"].idxmax()
df.loc[topper_index]
print("Topper name:",df.loc[topper_index,"StudentName"])
print("USN:",df.loc[topper_index,"StudentID"])
```

find the top scorer in mathematics.
print their (name,id and marks) in mathematics

```python
top_maths = df.loc[df["Mathematics"].idxmax()]
print("Name:", top_maths["StudentName"])
print("ID:", top_maths["StudentID"])
print("Mathematics Marks:", top_maths["Mathematics"])
```

print all the names whose total is above 400


```python
above_400 = df[df["Total"]>400]
print("Above 400 names:",above_400["StudentName"])
```


find the least average(name,id)


```python
least_avg = df.loc[df["Average"].idxmin()]
print("Name:", least_avg["StudentName"])
print("ID:", least_avg["StudentID"])
print("Average:", least_avg["Average"])
```
